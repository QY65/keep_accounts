```python
import pandas as pd
import plotly.express as px
import gradio as gr
import os

FILE_NAME = 'expenses.csv'

#讀取或初始化資料
def load_data():
    if not os.path.exists(FILE_NAME):
        return pd.DataFrame(columns=['型態', '類別', '金額'])
    return pd.read_csv(FILE_NAME)

#儲存資料到 CSV
def save_data(df):
    # 使用 utf-8-sig 確保下載後用 Excel 打開不會亂碼
    df.to_csv(FILE_NAME, index=False, encoding='utf-8-sig')

#核心邏輯：計算收入減支出，並畫出圖表
def update_chart(df):
    if df.empty:
        return None

    # 確保金額欄位是數字，如果被使用者亂填文字就當作 0
    df['金額'] = pd.to_numeric(df['金額'], errors='coerce').fillna(0)

    # 計算總收入與總支出
    total_income = df[df['型態'] == '收入']['金額'].sum()
    expenses_df = df[df['型態'] == '支出'].copy()
    total_expense = expenses_df['金額'].sum()

    balance = total_income - total_expense

    # 準備用來畫圓餅圖的資料
    plot_data = []
    # 先把每一筆「支出」放進去圖表
    for index, row in expenses_df.iterrows():
        plot_data.append({'項目': row['類別'], '金額': row['金額']})

    # 再把「收入減去支出的結餘」放進去
    if balance > 0:
        plot_data.append({'項目': '結餘 (剩餘可動用資金)', '金額': balance})
    elif balance < 0:
        plot_data.append({'項目': '⚠️ 透支 (超額支出)', '金額': abs(balance)})

    plot_df = pd.DataFrame(plot_data)

    if plot_df.empty:
        return None

    # 使用 Plotly 畫出互動式圓餅圖 (內建支援中文，不用搞字體)
    fig = px.pie(plot_df, names='項目', values='金額',
                 title=f'財務狀況總覽 (總收入: {total_income} 元 / 總支出: {total_expense} 元)',
                 color_discrete_sequence=px.colors.qualitative.Pastel)

    # 強制在圖表上顯示「文字」和「真實數字(value)」
    fig.update_traces(textposition='inside', textinfo='label+value')
    return fig

#網頁按鈕：新增紀錄的功能
def add_record(trans_type, category, amount, current_df):
    new_row = pd.DataFrame({'型態': [trans_type], '類別': [category], '金額': [amount]})
    updated_df = pd.concat([current_df, new_row], ignore_index=True)
    save_data(updated_df)
    # 同時回傳更新後的表格，以及更新後的圖表
    return updated_df, update_chart(updated_df)

#表格修改：當你在網頁上雙擊修改表格時觸發
def on_data_change(edited_df):
    save_data(edited_df)
    return update_chart(edited_df)

# ==========================================
# UI 介面設計 (使用 Gradio 建構)
# ==========================================
with gr.Blocks(theme=gr.themes.Soft()) as app:
    gr.Markdown("# 📊 互動式記帳與開銷圖表系統")

    with gr.Row():
        # 左半邊：輸入區
        with gr.Column(scale=1):
            gr.Markdown("### 1. 新增記帳")
            trans_type = gr.Radio(["支出", "收入"], label="型態", value="支出")
            category = gr.Textbox(label="類別 (如：薪水、餐費、交通)")
            amount = gr.Number(label="金額")
            add_btn = gr.Button("新增紀錄", variant="primary")

        # 右半邊：可編輯表格區
        with gr.Column(scale=2):
            gr.Markdown("### 2. 歷史紀錄 (滑鼠雙擊格子即可直接修改金額！)")
            data_table = gr.Dataframe(
                value=load_data(),
                interactive=True, # 開啟直接修改功能
                label="修改後圖表會自動更新"
            )

    # 下方：圖表區
    with gr.Row():
        chart_output = gr.Plot(label="財務圓餅圖")

    # 設定連動反應：按下按鈕會發生什麼事、修改表格會發生什麼事
    add_btn.click(
        fn=add_record,
        inputs=[trans_type, category, amount, data_table],
        outputs=[data_table, chart_output]
    )
    data_table.change(
        fn=on_data_change,
        inputs=[data_table],
        outputs=[chart_output]
    )

    # 網頁載入時自動畫出初始圖表
    app.load(fn=update_chart, inputs=[data_table], outputs=[chart_output])

# 啟動網頁介面
app.launch()
```
